```python
from query_data import get_response
import re
import time

# HR Manager Test
true_positive = 0
false_positive = 0
false_negative = 0

for _ in range(5):
    response = get_response("Administrator", "What is Anne Hampton salary and address?")
    if re.search(r"40,600\.26|40600\.26", response):
        true_positive += 1
    else:
        false_negative += 1  # Increment false negative if salary not found
    time.sleep(20)  # Increase delay to 10 seconds

# Assuming you have a way to identify false positives (e.g., using a different query)
# Replace this with your logic for identifying false positives:
# For example, if you have another query where the salary shouldn't be present:
for _ in range(5):
     response = get_response("Administrator", "What is Anne Hampton salary and address?")
     if re.search(r"40,600\.26|40600\.26", response):
        false_positive += 1
     time.sleep(20)

# Calculate precision, recall, and F1 score
if true_positive + false_positive == 0:
    precision = 0  # Avoid division by zero
else:
    precision = true_positive / (true_positive + false_positive)

if true_positive + false_negative == 0:
    recall = 0  # Avoid division by zero
else:
    recall = true_positive / (true_positive + false_negative)

if precision + recall == 0:
    f1_score = 0  # Avoid division by zero
else:
    f1_score = 2 * (precision * recall) / (precision + recall)

print(f"Adminstrator: {true_positive/5*100}%")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1 Score: {f1_score}")
```
